### Matrix multiplication

Two main ways of perming multiplicatiions in neural networks and deep learning are:

1. Element wise multiplication
2. Matrix multiplications (dot product)

There are two main rules that performing matrix multiplications need to satisfy:
1. The inner dimensions much match:
- `(3, 2) @ (3,2) wont work`
- `(2, 3) @ (3, 2) will work`
- `(3,2) @ (2,3) will work`
2. The resulting matrix has the shape of the outer dimensions:
- `(2, 3) @ (3,2) -> (2,2)`
- `(3,2) @ (2, 3) -> (3, 3)`

````python
print(x * y)
````

````python
print(x)
print(y)
print(torch.matmul(x, y))
````

````python
%%time
value = 0

for i in range(len(x)):
  value += x[i] + y[i]

value
````

````python
%%time
torch.matmul(x,y)
````

````python
# more mat muls


torch.matmul(torch.rand(3,2), torch.rand(2, 3))
````

## One of the most common errors in deep learning : shape errors

````python
tensor_A  = torch.tensor([[1,2], [3,4], [5,6]])
tensor_B = torch.tensor([[7,10], [8, 11], [9, 12]])

# torch.mm(tensor_A, torch_B)

````

````python
tensor_A.shape, tensor_B.shape
````

To fix our tensor shape issues, we can manipulate the shape of one of our tensors using a transpose. A transpose switches the axes or dimensions of a given tensor.

````python
tensor_B, tensor_B.T

tensor_B.shape, tensor_B.T.shape
````

````python
print(f'original shapes: tensor_a = {tensor_A.shape}, tensor_b = {tensor_B.shape}')
print(f'new shapes: tensor_a = {tensor_A.shape} (same shape as above), tensor_b.T = {tensor_B.T.shape}')
print(f'Multiplying: {tensor_A.shape} @ {tensor_B.T.shape} <- inner dimensions must match')
print('Output:\n')
output = torch.matmul(tensor_A, tensor_B.T)
print(output)
print(f'\n output shape: {output.shape}')
````

## Reshaping, stacking, squeezing and unsqueezing tensors

* Reshaping - reshaping an input tensor to a defined shape
* View - Returns a view of an input tensor of certain shape but keep the same memory as original tensor
* Stacking - combine multiple tensors on top of each other (vstack) and (hstack)
* Squeeze - remove all `1` dimensions from a tensor
* Unsqueeze - add a `1` dimension to a target tensor
* Permute - Return a view of the input with the dimensions permuted (swapped) in a certain way

````python
import torch
x = torch.arange(1., 10.)
x, x.shape, x.dtype
````

````python
# Add an extra dimension
x_rs = x.reshape(1, 9)
x_rs, x_rs.shape
````

````python
# change the view of th tensor

z = x.view(1, 9)
z, z.shape
````

````python
# changing z changes x because a view of a tensor shares the same mmory as the original
z[:, 0] = 5
z, x
````

````python
# Stack tensors on top of each other

x_stacked = torch.stack([x,x,x,x], dim=0)
x_stacked
````

````python
x_vstacked = torch.vstack([x,x,x,x])
x_vstacked
````

````python
x_hstacked = torch.hstack([x,x,x,x])
x_hstacked
````

````python
# torch.squeeze()
x_rs, x_rs.shape
````

````python
x_rs.squeeze(), x_rs.squeeze().shape
````

````python
# torch.unsqueeze()

x_squeezed = x_rs.squeeze()
x_squeezed, x_squeezed.shape
````

````python
x_usq = torch.unsqueeze(x_squeezed, dim=0)
x_usq.unsqueeze(dim=1), x_usq.shape
````

````python
x_perm = torch.permute(x_rs, (0, 1))
x_perm
````

## Reproducibility (trying to take random out of random)
In shor thow a neural networks learns:


`start with random numbers -> tensor operations -> update random numbers totry  and make them better representations of the data -> again -> again -> again..`

To reduce the randomness in neural networks and pytorch comes the concept of a **random seed**.


Essentially what the random seed does is "flavour" the randomness.            

````python
import torch


tena = torch.rand(3,4)
tenb = torch.rand(3,4)

print(tena)
print(tenb)
print(tena == tenb)
````

````python
# lets make some random but reproducible tensors

import torch

RANDOM_SEED = 42

torch.manual_seed(RANDOM_SEED)
tena = torch.rand(3,4)

torch.manual_seed(RANDOM_SEED)
tenb = torch.rand(3,4)

print(tena)
print(tenb)
print(tena == tenb)
````
